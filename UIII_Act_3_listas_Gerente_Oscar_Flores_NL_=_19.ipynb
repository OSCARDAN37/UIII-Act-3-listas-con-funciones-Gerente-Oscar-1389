# Lista Gerente


```python
gerentes = []

def agregar():
    id = input("ID: ")
    nombre = input("Nombre: ")
    depto = input("Departamento: ")

    gerente = [id, nombre, depto]
    gerentes.append(gerente)
    print("Gerente agregado")

def listar():
    if len(gerentes) == 0:
        print("No hay registros")
    else:
        for g in gerentes:
            print(g[0], g[1], g[2])

def buscar():
    id_buscar = input("ID a buscar: ")
    encontrado = False
    for g in gerentes:
        if g[0] == id_buscar:
            print("Encontrado:")
            print(g[0], g[1], g[2])
            encontrado = True
    if not encontrado:
        print("Gerente no encontrado")

while True:
    print("\n--- SISTEMA DE GERENTES ---")
    print("1. Agregar")
    print("2. Listar")
    print("3. Buscar")
    print("4. Salir")

    op = input("Opción: ")
    if op == "1":
        agregar()
    elif op == "2":
        listar()
    elif op == "3":
        buscar()
    elif op == "4":
        print("Saliendo...")
        break
    else:
        print("Opción inválida")
print("Programa realizado por Oscar Flores NL = 19")
```
