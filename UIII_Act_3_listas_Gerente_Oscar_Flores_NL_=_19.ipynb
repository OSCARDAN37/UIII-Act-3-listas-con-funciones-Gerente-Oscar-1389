# Lista Gerente


```python
class Gerente:
    lista = []

    def __init__(self, id, nombre, depto):
        self.id = id
        self.nombre = nombre
        self.depto = depto
while True:
    print("\n--- SISTEMA DE GERENTES ---")
    print("1. Agregar")
    print("2. Listar")
    print("3. Buscar")
    print("4. Salir")
    op = input("Opción: ")
    # Agregar gerente
    if op == "1":
        id = input("ID: ")
        nombre = input("Nombre: ")
        depto = input("Departamento: ")

        gerente = Gerente(id, nombre, depto)
        Gerente.lista.append(gerente)
        print("Gerente agregado")
    # Listar gerentes
    elif op == "2":
        if len(Gerente.lista) == 0:
            print("No hay registros")
        else:
            for g in Gerente.lista:
                print(g.id, g.nombre, g.depto)
    # Buscar gerente
    elif op == "3":
        buscar = input("ID a buscar: ")
        encontrado = False
        for g in Gerente.lista:
            if g.id == buscar:
                print("Encontrado:")
                print(g.id, g.nombre, g.depto)
                encontrado = True
        if not encontrado:
            print("Gerente no encontrado")
    # Salir
    elif op == "4":
        print("Saliendo...")
        break
    else:
        print("Opción inválida")
print("Programa realizado por Oscar Flores NL = 19")
```
